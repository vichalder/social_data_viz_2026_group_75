## Part 1

1. **Know Your Audience**: Tailor the complexity and detail of the plot to the specific background and needs of the people viewing it.
2. **Identify Your Message**: Every figure should have a clear, specific purpose or story it is intended to convey.
3. **Adapt to the Medium**: Design figures differently depending on whether they will appear in a printed journal, on a large poster, or in a presentation slide.
4. **Captions are Mandatory**: Ensure the figure and its caption are self-contained so the reader can understand the data without reading the full text.
5. **Avoid Default Settings**: Do not rely on software defaults; manually adjust scales, fonts, and colors to improve clarity.
6. **Use Color Purposefully**: Use color to highlight important data or distinguish categories, ensuring it remains accessible to those with color-blindness.
7. **Do Not Mislead**: Use honest scales and avoid unnecessary 3D effects or visual distortions that misrepresent the data.
8. **Avoid "Chartjunk"**: Remove unnecessary elements like heavy grid lines, borders, and decorations that do not represent actual data.
9. **Message Trumps Beauty**: While aesthetics are important, the clarity of the information and the message must always take priority.
10. **Get Feedback**: Show your plots to others to verify that they interpret the data and the message as you intended.
