### Let us again merge similar categories

```python
merged_data['Incident Category'].unique()
```

```python
category_fine_tuning = {
    # Spelling & Punctuation cleanup
    'Motor Vehicle Theft?': 'Motor Vehicle Theft',
    'Weapons Offence': 'Weapons Offense',
    'Weapons Carrying Etc': 'Weapons Offense',
    
    # User-specified Human Trafficking mapping
    'Human Trafficking (A), Commercial Sex Acts': 'Prostitution',
    'Human Trafficking, Commercial Sex Acts': 'Prostitution',
    'Human Trafficking (B), Involuntary Servitude': 'Rape',
    
    # Naming convention overlap
    'Drug Violation': 'Drug Offense',
    'Suspicious': 'Suspicious Occ',
    'Malicious Mischief': 'Vandalism',
    
    # Miscellaneous consolidation
    'Other Miscellaneous': 'Other Offenses',
    'Other': 'Other Offenses'
}
```

```python
merged_data['Incident Category'] = merged_data['Incident Category'].replace(category_fine_tuning)
```

```python
print(merged_data['Incident Datetime'].dt.year.unique())
```

### Only include full years

```python
# 1. Get the absolute latest timestamp in the data
latest_timestamp = merged_data['Incident Datetime'].max()

# 2. Define what the "end of that year" would be
# This creates a timestamp for Dec 31st of that year
end_of_year = pd.Timestamp(year=latest_timestamp.year, month=12, day=31)

# 3. Logic: If the latest data point is BEFORE the end of its year, 
# it's an incomplete year.
if latest_timestamp < end_of_year:
    df_filtered = merged_data[merged_data['Incident Datetime'].dt.year < latest_timestamp.year]
else:
    # If the data goes all the way to Dec 31st, keep everything
    df_filtered = merged_data.copy()
```
